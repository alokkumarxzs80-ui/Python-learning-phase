```python
'''We are building a billing system for a local cafe. Your task is to write the Python code for a MenuItem class.

Requirements:

Create a class named MenuItem.

Write an __init__ method that accepts self, name, price, and category (e.g., Beverage, Snack).

Write a method called get_info(self) that returns a formatted string like: "Masala Chai (Beverage): ₹40".

Outside the class, create two objects from your blueprint: one for a ₹40 Masala Chai, and one for a ₹120 Samosa Chat.

Print the result of get_info() for both objects.'''

class MenuItem:
  def __init__(self, name, price, category):
    self.name = name
    self.price = price
    self.category = category

  def get_info(self):
    return f"{self.name} ({self.category}): ₹{self.price}"

# Outside the class (unindented)
cart1 = MenuItem("Masala Chai", "40", "Beverage")
cart2 = MenuItem("Samosa Chat", "120", "Snack")

print(cart1.get_info())
print(cart2.get_info())
```
